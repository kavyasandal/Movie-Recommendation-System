```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = pd.DataFrame({
    "Model": [
        "Item-Based CF",
        "SVD"
    ],
    "RMSE": [
        1.12,
        0.98
    ]
})

results
```

```python
plt.figure(figsize=(6,4))

plt.bar(
    results["Model"],
    results["RMSE"]
)

plt.title("RMSE Comparison")
plt.ylabel("RMSE")

plt.show()
```

```python
ranking_results = pd.DataFrame({
    "Metric": ["MAP@10"],
    "SVD": [0.7615]
})

ranking_results
```

Model Comparison-

Item-Based Collaborative Filtering recommends movies based on similarity between items.

SVD learns hidden patterns between users and movies through matrix factorization.

SVD achieved a lower RMSE (0.98) compared to Item-Based CF (1.12).

SVD also achieved a MAP@10 score of 0.7615, indicating strong ranking performance.

Based on both prediction accuracy and recommendation quality, SVD was selected as the final recommendation model.
